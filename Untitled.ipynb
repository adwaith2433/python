```python
from abc import ABC,abstractmethod
class bank(ABC):
    @abstractmethod
    def interest(self):
        pass
class sbi(bank):
    def interest(self):
        print("interest rate:6.5%")
class hdfc(bank):
    def interest(self):
        print("interest rate:7%")
s=sbi()
s.interest()
h=hdfc()
h.interest()
```

```python
from abc import ABC,abstractmethod
class vehicle(ABC):
    @abstractmethod
    def start(self):
        pass
class car(vehicle):
    def start(self):
        print("car starts with a key")
class bike(vehicle):
    def start(self):
        print("bike starts with a self-start button")
c=car()
c.start()
b=bike()
b.start()
```

```python

def atm():
    balance = 1000
    pin = "1234"
    print("Welcome to the ATM!")
    entered_pin = input("Enter your 4-digit PIN: ")

    if entered_pin != pin:
        print("Incorrect PIN. Exiting.")
        return

    while true:
        print("\nATM Menu:")
        print("1. Check Balance")
        print("2. Deposit")
        print("3. Withdraw")
        print("4. Exit")

        choice = input("Enter your choice (1-4): ")

        if choice == '1':
            print(f"Your current balance is: ${balance}")

        elif choice == '2':
            amount_str = input("Enter amount to deposit: $")

            if amount_str.isdigit():
                amount = int(amount_str)

                if amount > 0:
                    balance += amount
                    print(f"Successfully deposited ${amount}. New balance: ${balance}")
                else:
                    print("Deposit amount must be positive.")
            else:
                print("Invalid input. Please enter a number.")

        elif choice == '3':
            amount_str = input("Enter amount to withdraw: $")

            if amount_str.isdigit():
                amount = int(amount_str)

                if amount > 0:
                    if amount <= balance:
                        balance -= amount
                        print(f"Successfully withdrew ${amount}. New balance: ${balance}")
                    else:
                        print("Insufficient funds.")
                else:
                    print("Withdrawal amount must be positive.")
            else:
                print("Invalid input. Please enter a number.")

        elif choice == '4':
            print("Thank you for using the ATM. Goodbye!")
            break

        else:
            print("Invalid choice. Please select a valid option (1-4).")


atm()
```

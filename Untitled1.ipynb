```python
accounts={
    "1001":{"pin":"1234","name":"rahul","balance":15000},
    "1002":{"pin":"5678","name":"priya","balance":10000},
    "1003":{"pin":"4321","name":"arun","balance":20000}
}
def login():
    account=input("enter account number:")
    pin=input("enter pin:")
    if account in accounts and accounts[account]["pin"]==pin:
       print("\nlogin successful!")
       return account
    else:
        print("\nInvalid Account Number or PIN!")
        return None
def balance(account):
      print(f"\nCurrent Balance: ₹{accounts[account]['balance']}")
def deposit(account):
    amount = float(input("Enter amount to deposit: ₹"))
    if amount > 0:
        accounts[account]["balance"] += amount
        print("Deposit Successful!")
        print(f"Updated Balance: ₹{accounts[account]['balance']}")
    else:
        print("Invalid Amount!")
def withdraw(account):
    amount = float(input("Enter amount to withdraw: ₹"))
    if amount <= accounts[account]["balance"]:
        accounts[account]["balance"] -= amount
        print("Withdrawal Successful!")
        print(f"Remaining Balance: ₹{accounts[account]['balance']}")
    else:
        print("Insufficient Balance!")
def change_pin(account):
    old_pin=input("enter current pin:")
    if old_pin==accounts[account]["pin"]:
        new_pin=input("enter new pin:")
        confirm=input("confirm new pin:")
        if new_pin==confirm:
            accounts[account]["pin"]=new_pin
            print("pin changed successfully!")
        else:
            print("pins do not match!")
    else:
        print("incorrect current pin!")
def menu(account):
    while True:
        print("\n========== ATM MENU ==========")
        print("1.balance inquiry")
        print("2.deposit")
        print("3.withdraw")
        print("4.change pin")
        print("5.exit")
        choice=input("enter your choice:")
        if choice=="1":
            balance(account)
        elif choice=="2":
            deposit(account)
        elif choice=="3":
              withdraw(account)
        elif choice=="4":
            change_pin(account)
        elif choice=="5":
            print("\nThank you for using our ATM.")
            break
        else:
            print("invalid choice!")
print("================================")
print("atm management system")
print("================================")
user=login()
if user:
     print(f"\nWelcome, {accounts[user]['name']}!")
     menu(user)
```

```python
admin_username = "admin"
admin_password = "1234"
books = {
    101: {"title": "Python Programming", "author": "Guido", "available": True},
    102: {"title": "Data Structures", "author": "Mark", "available": True},
    103: {"title": "Database Systems", "author": "Thomas", "available": True}
}
students = {}
def admin_login():
    username = input("Enter Admin Username: ")
    password = input("Enter Password: ")
    if username == admin_username and password == admin_password:
        print("\nLogin Successful!")
        return True
    else:
        print("Invalid Username or Password")
        return False
def add_book():
    book_id = int(input("Enter Book ID: "))
    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")
    books[book_id] = {
        "title": title,
        "author": author,
        "available": True
    }
    print("Book Added Successfully!")
def remove_book():
    
    book_id = int(input("Enter Book ID: "))
    if book_id in books:
        del books[book_id]
        print("Book Removed Successfully!")
    else:
        print("Book Not Found!")
def search_book():
    title = input("Enter Book Title: ").lower()
    found = False
    for book_id, book in books.items():
        if title in book["title"].lower():
            print("\nBook ID:", book_id)
            print("Title:", book["title"])
            print("Author:", book["author"])
            print("Available:", book["available"])
            found = True
    if not found:
        print("Book Not Found!")
def issue_book():
    student = input("Enter Student Name: ")
    book_id = int(input("Enter Book ID: "))
    if book_id in books and books[book_id]["available"]:
        books[book_id]["available"] = False
        students[student] = book_id
        print("Book Issued Successfully!")
    else:
        print("Book Not Available!")
def return_book():
    student = input("Enter Student Name: ")
    if student in students:
        book_id = students[student]
        days = int(input("Enter Number of Days Used: "))
        fine = 0
        if days > 7:
            fine = (days - 7) * 5
        books[book_id]["available"] = True
        del students[student]
        print("Book Returned Successfully!")
        print("Fine: ₹", fine)
    else:
        print("No Book Issued!")
def student_records():
    if not students:
        print("No Records Found")
    else:
        for student, book_id in students.items():
            print(student, "->", books[book_id]["title"])
def display_books():
    print("\nAvailable Books")
    print("-" * 40)
    for book_id, book in books.items():
        print(book_id, "|", book["title"], "|", book["author"], "|", book["available"])
def menu():
    while True:
        print("\n====== LIBRARY MANAGEMENT SYSTEM ======")
        print("1. Add Book")
        print("2. Remove Book")
        print("3. Search Book")
        print("4. Issue Book")
        print("5. Return Book")
        print("6. Student Records")
        print("7. Display Books")
        print("8. Exit")
        choice = input("Enter Choice: ")
        if choice == "1":
            add_book()
        elif choice == "2":
            remove_book()
        elif choice == "3":
            search_book()
        elif choice == "4":
            issue_book()
        elif choice == "5":
            return_book()
        elif choice == "6":
            student_records()
        elif choice == "7":
            display_books()
        elif choice == "8":
            print("Thank You!")
            break
        else:
            print("Invalid Choice")
print("===== LIBRARY MANAGEMENT SYSTEM =====")

if admin_login():
    menu()
```

```python
import matplotlib.pyplot as plt
import numpy as np
branch=["it","ece","cse"]
it=[10,28,3,6,20]
ece=[30,8,18,7,12]
cse=[32,6,27,28,20]
a=np.arange(len(it))
barwidth=0.24
plt.bar(a-barwidth,it,color="#ff00ff",width=barwidth,edgecolor="#0000ff",label="it")
plt.bar(a,ece,color="#8000ff",width=barwidth,edgecolor="#00ffff",label='ece')
plt.bar(a+barwidth,cse,color="#ff0000",width=barwidth,edgecolor="#000",label='cse')
plt.xlabel("branch")
plt.ylabel("students passed")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
students=["arun","jobin","joe"]
#subjects=["malayalam","english","hindi"]
arun=[12,18,25]
jobin=[28,33,36]
jeo=[26,31,34]
a=np.arange(len(arun))
barwidth=0.25
plt.bar(a-barwidth,arun,color='r',width=barwidth,edgecolor='grey',label="arun")
plt.bar(a,jobin,color='g',width=barwidth,edgecolor='grey',label='jobin')
plt.bar(a+barwidth,jeo,color='b',width=barwidth,edgecolor='grey',label='cse')
plt.xlabel("students")
plt.ylabel("students marks")
plt.legend()
plt.show()
```
